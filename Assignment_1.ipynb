
6. How do conditional statements work in Python? Illustrate with examples?
->Conditional statements in Python allow the program to make decisions based on whether a condition is True or False. They control the flow of execution by executing code decision based on some precondition
examples are: if, if else, if elif else,nested if else.

```python
a=100
if a>0:
  print("the number is greater than 0")
```

```python
age=17
if age >=18:
    print("you will be served alcohol")
else:
    print("get out of the bar")
```

```python
a=10
if a>100:
    print("this block will be executed if a is greater than 100")
elif a<100:
    print("this block will be executed if a is lesser than 100")
else:
    print("the number is equal to 100")
```

```python
x=9
y=6
if x>5:
    if y>5:
        print("both x and y is greater than 5")
    else:
        print("x is greater than 5 but y is less than 5")
else:
    print("x is not greater than 5")
```
